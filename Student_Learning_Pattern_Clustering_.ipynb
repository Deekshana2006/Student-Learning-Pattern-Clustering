Part 4– Data Visualization

create a histogram

```python
import matplotlib.pyplot as plt

# Single column
plt.figure(figsize=(7,5))
plt.hist(df["ExamScore"], bins=10, color="#4C72B0", edgecolor="black")
plt.title("Histogram of ExamScore")
plt.xlabel("ExamScore")
plt.ylabel("Frequency")
plt.show()
```

```python
# All numeric columns at once (grid of histograms)
df.hist(figsize=(15, 12), bins=10, color="#4C72B0", edgecolor="black")
plt.tight_layout()
plt.show()
```

Create a Box Plot

```python
import matplotlib.pyplot as plt

# Single column
plt.figure(figsize=(5,6))
plt.boxplot(df["ExamScore"], vert=True)
plt.title("Box Plot of ExamScore")
plt.ylabel("ExamScore")
plt.show()
```

```python
# All numeric columns at once
df.plot(kind="box", figsize=(15,6), rot=45)
plt.title("Box Plot of All Numeric Columns")
plt.tight_layout()
plt.show()
```

Create a Bar Chart

```python
import matplotlib.pyplot as plt

avg_by_gender = df.groupby("Gender")["ExamScore"].mean()

plt.figure(figsize=(6,5))
avg_by_gender.plot(kind="bar", color="#55A868", edgecolor="black")
plt.title("Average ExamScore by Gender")
plt.xlabel("Gender")
plt.ylabel("Average ExamScore")
plt.xticks(rotation=0)
plt.show()
```

```python
# Alternative: count of students by LearningStyle
counts_by_style = df["LearningStyle"].value_counts().sort_index()

plt.figure(figsize=(6,5))
counts_by_style.plot(kind="bar", color="#4C72B0", edgecolor="black")
plt.title("Number of Students by LearningStyle")
plt.xlabel("LearningStyle")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

create a scatter plot


```python

plt.figure(figsize=(6,5))
plt.scatter(df["StudyHours"], df["ExamScore"], color="yellow", alpha=0.4)
plt.title("StudyHours vs ExamScore")
plt.xlabel("StudyHours")
plt.ylabel("ExamScore")
plt.show()
```

```python
# Attendance vs ExamScore
plt.figure(figsize=(6,5))
plt.scatter(df["Attendance"], df["ExamScore"], color="purple", alpha=0.4)
plt.title("Attendance vs ExamScore")
plt.xlabel("Attendance")
plt.ylabel("ExamScore")
plt.show()
```

```python
# AssignmentCompletion vs ExamScore
plt.figure(figsize=(6,5))
plt.scatter(df["AssignmentCompletion"], df["ExamScore"], color="green", alpha=0.4)
plt.title("AssignmentCompletion vs ExamScore")
plt.xlabel("AssignmentCompletion")
plt.ylabel("ExamScore")
plt.show()
```

Part 5– Correlation

calculate the correlation

```python
correlation_matrix = df.corr(numeric_only=True)
print(correlation_matrix)
```

```python


plt.figure(figsize=(10,8))
plt.imshow(correlation_matrix, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")
plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns, rotation=45, ha="right")
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
